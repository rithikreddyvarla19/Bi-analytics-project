# 03 KPI Reporting\n\nThis notebook calculates stakeholder-facing KPIs for the synthetic child welfare support services dataset.

```python
from pathlib import Path\nimport json\nimport pandas as pd\n\nPROJECT_ROOT = Path('..').resolve()\nDATA_PATH = PROJECT_ROOT / 'data' / 'processed' / 'child_welfare_cases_clean.csv'\nKPI_PATH = PROJECT_ROOT / 'data' / 'processed' / 'kpi_summary.json'\ndf = pd.read_csv(DATA_PATH, parse_dates=['referral_date'])
```

```python
with open(KPI_PATH, 'r', encoding='utf-8') as f:\n    kpis = json.load(f)\n\nsummary = pd.DataFrame([\n    {'KPI': 'Total cases', 'Value': kpis['total_cases']},\n    {'KPI': 'High-risk case percentage', 'Value': f"{kpis['high_risk_case_percentage']}%"},\n    {'KPI': 'Average days to first contact', 'Value': kpis['average_days_to_first_contact']},\n    {'KPI': 'Follow-up completion rate', 'Value': f"{kpis['follow_up_completion_rate']}%"},\n    {'KPI': 'Overall closure rate', 'Value': f"{kpis['overall_closure_rate']}%"},\n])\nsummary
```

```python
pd.Series(kpis['cases_by_state']).sort_values(ascending=False)
```

```python
pd.Series(kpis['cases_by_referral_source']).sort_values(ascending=False)
```

```python
pd.Series(kpis['service_utilization_by_type']).sort_values(ascending=False)
```

```python
pd.Series(kpis['closure_rate_by_risk_level']).sort_index()
```

## Reporting Interpretation\n\nThe KPI set is designed for operational review. Volume metrics show demand, risk metrics support prioritization, timeliness metrics assess responsiveness, follow-up metrics measure continuity, and outcome metrics support program evaluation.
